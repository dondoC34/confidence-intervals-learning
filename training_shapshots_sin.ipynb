# Training Snapshots — PyTorch Prediction Interval Neural Network (Sinusoidal)

This notebook generates a heteroscedastic sinusoidal dataset via `src.data`, trains `PiNeuralNetwork` while collecting **50 snapshots** across training, and visualizes the evolution of the **Upper Bound**, **Lower Bound**, and **Regression** curves in a **5-column x 10-row** grid.

```python
import numpy as np
import torch
import matplotlib.pyplot as plt

from src.data.dataset_generators import DatasetGenerator
from src.model.lr_schedule import TimeDecreasingLr
from src.model.pi_neural_network import PiNeuralNetwork
from src.plot.training_plots import TrainingSnapshot, plot_training_snapshots_grid
```

```python
# Reproducibility
SEED = 42
np.random.seed(SEED)
torch.manual_seed(SEED)

# 1. Generate synthetic dataset (sinusoid with quadratic Gaussian noise)
X_MIN, X_MAX = -2.0, 2.0
dataset_gen = DatasetGenerator(
    x_min=X_MIN,
    x_max=X_MAX,
    sample_size=500,
    validation_split=0.2,
    seed=SEED,
)
x_train, y_train, x_val, y_val = dataset_gen.generate_noisy_sin_gaussian(amplitude=1.5)
x_grid = np.linspace(X_MIN, X_MAX, 200)

# 2. Instantiate the PyTorch model
model = PiNeuralNetwork(
    input_dim=1,
    hidden_layers=[(100, "relu"), (50, "relu")],
    confidence=0.95,
    softening_factor=120.0,
    lr=1e-3,
    alpha=5.0,
    lambda_=0.02,
    mse_inactivity_epochs=50,
    lr_schedule=TimeDecreasingLr(decay=1e-4),
    verbose=False,
)

# 3. Train while collecting 50 snapshots (epoch 0 + 49 steps of 3 epochs = 147 epochs)
N_SNAPSHOTS = 50
EPOCHS_PER_STEP = 5

snapshots: list[TrainingSnapshot] = []
ubs0, lbs0, regr0 = model.predict(x_grid)
snapshots.append(TrainingSnapshot(epoch=0, ubs=ubs0, lbs=lbs0, regr=regr0))

for _ in range(N_SNAPSHOTS - 1):
    model.fit(x_train, y_train, epochs=EPOCHS_PER_STEP, batch_size=32)
    ubs, lbs, regr = model.predict(x_grid)
    snapshots.append(TrainingSnapshot(epoch=model.epoch, ubs=ubs, lbs=lbs, regr=regr))

# Final validation metrics
val_ubs, val_lbs, val_regr = model.predict(x_val)
covered = (y_val >= val_lbs) & (y_val <= val_ubs)
print(f"Device: {model.device} | Total epochs: {model.epoch}")
print(f"Validation PICP (target 0.95): {np.mean(covered):.3f}")
print(f"Validation MPIW (mean width): {np.mean(val_ubs - val_lbs):.4f}")
print(f"Validation Regression MSE: {np.mean((val_regr - y_val) ** 2):.4f}")
print(f"Inverted bounds: {int(np.sum(val_lbs > val_ubs))}")
```

```python
# 4. Plot 5 columns x 10 rows of training progression
y_margin = 0.15 * (y_train.max() - y_train.min())
fig = plot_training_snapshots_grid(
    snapshots=snapshots,
    x=x_train,
    y=y_train,
    x_grid=x_grid,
    ncols=5,
    nrows=10,
    x_lim=(X_MIN, X_MAX),
    y_lim=(float(y_train.min() - y_margin), float(y_train.max() + y_margin)),
    suptitle="PI-NN Training Progression — Sinusoidal (50 Snapshots: 5 cols x 10 rows)",
)
plt.show()
```
